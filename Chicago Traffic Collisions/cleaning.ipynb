```python
import pandas as pd

traffic_df = pd.read_csv("chicago_traffic_crashes_export.csv")
traffic_df.drop(columns = ["CRASH_DATE_EST_I", "SEC_CONTRIBUTORY_CAUSE", "PHOTOS_TAKEN_I", "LANE_CNT", "DATE_POLICE_NOTIFIED",
    "STATEMENTS_TAKEN_I", "DOORING_I", "WORK_ZONE_I", "WORK_ZONE_TYPE", "WORKERS_PRESENT_I", "LOCATION", "IDOT_CONTROL_NO"], 
    inplace = True)
traffic_df[["INTERSECTION_RELATED_I", "NOT_RIGHT_OF_WAY_I","HIT_AND_RUN_I"]].fillna("N")

traffic_df.to_csv("chicago_traffic_crashes_clean.csv", index=False)
```
